```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.colors import TwoSlopeNorm
import cartopy.mpl.ticker as cticker

# ----------------------------
# Settings
# ----------------------------
TIME0 = "2004-01-01"
TIME1 = "2020-12-31"
ZMIN  = 0
ZMAX  = 1000

PLOT_WHAT = "mean"   # "mean" (mmol m^-3 yr^-1) recommended; or "integral"
PCTL = 98            # percentile for color limits (95/98/99)

# Stippling settings (GOBAI only)
ADD_STIPPLING = True
STIPPLE_UPPER_PCT = 10  # upper 10% relative uncertainty

# ----------------------------
# Helper masks
# ----------------------------
def mask_mediterranean(lat, lon):
    lat_min, lat_max = 30, 46
    lon_min, lon_max = -6, 37
    return (lat >= lat_min) & (lat <= lat_max) & (lon >= lon_min) & (lon <= lon_max)

# ----------------------------
# Load datasets
# ----------------------------
ds  = xr.load_dataset("data/CONTENT_DIC_v2-1.nc")          # GOBAI-like grid with oxy + uncer
ito = xr.load_dataset("data/o2_NN_ship+Argo_ensmean.nc")   # ML4O2 oxygen only

# ============================================================
# 1) GOBAI: trend over 200–1000 dbar, 2005–2020 (PER YEAR)
# ============================================================
g = (
    ds
    .sel(time=slice(TIME0, TIME1))
    .sel(pres=slice(ZMIN, ZMAX))
)

# numeric time axis in years since start (robust)
t_year_g = ((g["time"] - g["time"].isel(time=0)) / np.timedelta64(1, "D")) / 365.25
g = g.assign_coords(t_year=("time", t_year_g.data))

# slope per year (mmol m^-3 yr^-1)
fit_g = g["oxy"].polyfit(dim="t_year", deg=1, skipna=True)
o2_trend_g = fit_g["polyfit_coefficients"].sel(degree=1)

# thickness weights (dbar)
dp = g["pres"].diff("pres").fillna(0)
dp = dp.broadcast_like(o2_trend_g)

# integrated and mean trends
o2_trend_int_gobai  = (o2_trend_g * dp).sum("pres", skipna=True).rename("O2_trend_int_200_1000")
o2_meantrend_gobai  = ((o2_trend_g * dp).sum("pres", skipna=True) / dp.sum("pres")).rename("O2_meantrend_200_1000")

print("GOBAI ready. Mean range:",
      float(o2_meantrend_gobai.min()), float(o2_meantrend_gobai.max()))

# ============================================================
# 2) ML4O2: trend over 200–1000 m, 2005–2020 (PER YEAR)
# ============================================================
i = (
    ito
    .sel(time=slice(TIME0, TIME1))
    .sel(depth=slice(ZMIN, ZMAX))
)

t_year_i = ((i["time"] - i["time"].isel(time=0)) / np.timedelta64(1, "D")) / 365.25
i = i.assign_coords(t_year=("time", t_year_i.data))

fit_i = i["o2"].polyfit(dim="t_year", deg=1, skipna=True)
o2_trend_i = fit_i["polyfit_coefficients"].sel(degree=1)  # mmol m^-3 yr^-1

dz = i["depth"].diff("depth").fillna(0)
dz = dz.broadcast_like(o2_trend_i)

o2_trend_int_ito = (o2_trend_i * dz).sum("depth", skipna=True).rename("O2_trend_int_200_1000")
o2_meantrend_ito = ((o2_trend_i * dz).sum("depth", skipna=True) / dz.sum("depth")).rename("O2_meantrend_200_1000")

print("ML4O2 ready. Mean range:",
      float(o2_meantrend_ito.min()), float(o2_meantrend_ito.max()))

# ============================================================
# 3) Choose what to plot
# ============================================================
if PLOT_WHAT == "mean":
    gobai_field = o2_meantrend_gobai
    ito_field   = o2_meantrend_ito
    cbar_label  = r"Thickness-weighted mean O$_2$ trend (mmol m$^{-3}$ yr$^{-1}$)"
    title_suffix = "mean"
elif PLOT_WHAT == "integral":
    gobai_field = o2_trend_int_gobai
    ito_field   = o2_trend_int_ito
    cbar_label  = r"Vertically integrated O$_2$ trend (product-specific thickness units yr$^{-1}$)"
    # NOTE: GOBAI integrates with dp (dbar) and ML4O2 with dz (m).
    # For cross-product amplitude comparison of integrals, convert dp -> meters (optional).
    title_suffix = "integrated"
else:
    raise ValueError("PLOT_WHAT must be 'mean' or 'integral'.")

# Subset ML4O2 to GOBAI latitude range for apples-to-apples view
lat_g = ds["lat"].values
ito_field_sub = ito_field.sel(lat=slice(lat_g.min(), lat_g.max()))

# ============================================================
# 4) GOBAI-only stippling mask: upper 10% relative O2 uncertainty
#     (computed over the same time/depth window, thickness-weighted)
# ============================================================
high_unc_mask = None
if ADD_STIPPLING:
    # thickness-weighted mean O2 and uncer for each time step
    o2_bar_t  = (g["oxy"]   * dp).sum("pres", skipna=True) / dp.sum("pres")
    unc_bar_t = (g["uncer"] * dp).sum("pres", skipna=True) / dp.sum("pres")

    # stable time-mean relative uncertainty metric (%)
    o2_bar  = o2_bar_t.mean("time", skipna=True)
    unc_bar = unc_bar_t.mean("time", skipna=True)

    eps = np.finfo(float).eps
    rel_unc = (unc_bar / (np.abs(o2_bar) + eps)) * 100.0

    # upper 10% threshold over finite ocean points
    rel_vals = rel_unc.values
    thr = np.nanpercentile(rel_vals[np.isfinite(rel_vals)], 100 - STIPPLE_UPPER_PCT)
    high_unc_mask = (rel_unc > thr)

    # ensure stippling only where trend map exists
    high_unc_mask = high_unc_mask.where(np.isfinite(gobai_field))

    print(f"GOBAI stippling threshold (relative uncertainty, %): {float(thr)}")

# ============================================================
# 5) Prepare grids + Mediterranean mask
# ============================================================
lon_g = gobai_field["lon"].values
lat_g = gobai_field["lat"].values
lon2d_g, lat2d_g = np.meshgrid(lon_g, lat_g)

lon_i = ito_field_sub["lon"].values
lat_i = ito_field_sub["lat"].values
lon2d_i, lat2d_i = np.meshgrid(lon_i, lat_i)

gobai_map = gobai_field.values
ito_map   = ito_field_sub.values

med_g = mask_mediterranean(lat2d_g, lon2d_g)
med_i = mask_mediterranean(lat2d_i, lon2d_i)

gobai_map = np.where(med_g, np.nan, gobai_map)
ito_map   = np.where(med_i, np.nan, ito_map)

# apply Mediterranean mask to stippling too (panel a only)
if ADD_STIPPLING and (high_unc_mask is not None):
    high_unc_mask = high_unc_mask.where(~xr.DataArray(med_g, dims=("lat","lon"),
                                                      coords={"lat": gobai_field["lat"], "lon": gobai_field["lon"]}))

# robust symmetric color limits
allvals = np.concatenate([gobai_map.ravel(), ito_map.ravel()])
allvals = allvals[np.isfinite(allvals)]
v = np.nanpercentile(np.abs(allvals), PCTL)
norm = TwoSlopeNorm(vcenter=0, vmin=-v, vmax=v)
print(f"Using symmetric color range ± {v}")

# ============================================================
# 6) Plot (2 rows: GOBAI top, ML4O2 bottom)
# ============================================================
fig = plt.figure(figsize=(13, 12), dpi=300)

land = cfeature.NaturalEarthFeature(
    "physical", "land", "50m",
    edgecolor="none", facecolor="lightgrey"
)

# -------------------------------------------------
# Panel (a): GOBAI (top row)
# -------------------------------------------------
ax1 = fig.add_subplot(2, 1, 1, projection=ccrs.Robinson())
ax1.set_global()
ax1.add_feature(land)
ax1.text(-0.05, 1.05, "a)", transform=ax1.transAxes, fontsize=12, va="top")

gl1 = ax1.gridlines(draw_labels=True, alpha=0.3, linestyle="--")
gl1.top_labels = True
gl1.bottom_labels = False
gl1.left_labels = True
gl1.right_labels = True
gl1.xformatter = cticker.LongitudeFormatter()
gl1.yformatter = cticker.LatitudeFormatter()

fig.canvas.draw()
for lbl in gl1.xlabel_artists + gl1.ylabel_artists:
    lbl.set_fontsize(6)

im1 = ax1.pcolormesh(
    lon2d_g, lat2d_g, gobai_map,
    transform=ccrs.PlateCarree(),
    cmap="RdBu_r", norm=norm
)

# Stippling (GOBAI only)
if ADD_STIPPLING and (high_unc_mask is not None):
    ax1.contourf(
        lon2d_g, lat2d_g, high_unc_mask.values.astype(float),
        levels=[0.5, 1.5],
        colors="none",
        hatches=["....."],
        transform=ccrs.PlateCarree()
    )

ax1.set_title(
    f"GOBAI O$_2$ trend (0–1000 m, 2005–2020) [{title_suffix}]",
    fontsize=10
)

# -------------------------------------------------
# Panel (b): ML4O2 (bottom row)
# -------------------------------------------------
ax2 = fig.add_subplot(2, 1, 2, projection=ccrs.Robinson())
ax2.set_global()
ax2.add_feature(land)
ax2.text(-0.05, 1.05, "b)", transform=ax2.transAxes, fontsize=12, va="top")

gl2 = ax2.gridlines(draw_labels=True, alpha=0.3, linestyle="--")
gl2.top_labels = False
gl2.bottom_labels = True
gl2.left_labels = True
gl2.right_labels = True
gl2.xformatter = cticker.LongitudeFormatter()
gl2.yformatter = cticker.LatitudeFormatter()

fig.canvas.draw()
for lbl in gl2.xlabel_artists + gl2.ylabel_artists:
    lbl.set_fontsize(6)

im2 = ax2.pcolormesh(
    lon2d_i, lat2d_i, ito_map,
    transform=ccrs.PlateCarree(),
    cmap="RdBu_r", norm=norm
)

ax2.set_title(
    f"ML4O2 O$_2$ trend (0–1000 m, 2004–2020) [{title_suffix}]",
    fontsize=10
)

# -------------------------------------------------
# Shared colorbar (bottom)
# -------------------------------------------------
cax = fig.add_axes([0.25, 0.06, 0.5, 0.018])
cbar = fig.colorbar(im1, cax=cax, orientation="horizontal", extend="both")
cbar.set_label(cbar_label)

# -------------------------------------------------
# Final layout & save
# -------------------------------------------------
plt.subplots_adjust(hspace=0.25)

outpath = (
    f"figs/FigS_O2_robustness_GOBAI_vs_ML4O2_{title_suffix}_2rows_stippling.png"
    if ADD_STIPPLING else
    f"figs/FigS_O2_robustness_GOBAI_vs_ML4O2_{title_suffix}_2rows.png"
)

fig.savefig(outpath, bbox_inches="tight", dpi=300)
plt.show()

print("Saved:", outpath)

```

```python
# ============================================================
# FIGURE: Sign disagreement map with GOBAI uncertainty coloring
# Black = sign disagreement in low/moderate GOBAI uncertainty
# Red   = sign disagreement in high GOBAI uncertainty (upper 10%)
#
# PLUS: prints summary stats:
#   - total valid pixels (both trends finite, outside Mediterranean)
#   - % agree / % disagree
#   - among disagree, % in high GOBAI uncertainty
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import cartopy.mpl.ticker as cticker

# ----------------------------
# Settings (match your other fig)
# ----------------------------
TIME0, TIME1 = "2004-01-01", "2020-12-31"
ZMIN, ZMAX   = 0, 1000
PCTL_UNC     = 90   # upper 10% threshold
DOTSIZE      = 4

# Mediterranean mask (same as before)
def mask_mediterranean(lat, lon):
    return (lat >= 30) & (lat <= 46) & (lon >= -6) & (lon <= 37)

# ----------------------------
# 1) Regrid ML4O2 onto GOBAI grid
# ----------------------------
ito_on_gobai = o2_meantrend_ito.interp(
    lat=o2_meantrend_gobai["lat"],
    lon=o2_meantrend_gobai["lon"],
    method="nearest"
)

gtrend = o2_meantrend_gobai
itrend = ito_on_gobai

valid = np.isfinite(gtrend) & np.isfinite(itrend)

# Opposite signs => disagreement
sign_disagree = ((np.sign(gtrend) * np.sign(itrend)) < 0).where(valid)

# ----------------------------
# 2) Compute GOBAI high-uncertainty mask (upper 10%)
#    Use mean O2 field (NOT trend) to avoid division-by-zero blowups
# ----------------------------
g_unc = (
    ds
    .sel(time=slice(TIME0, TIME1))
    .sel(pres=slice(ZMIN, ZMAX))
)

dp = g_unc["pres"].diff("pres").fillna(0)

# thickness-weighted mean O2 and mean uncertainty at each time
o2_bar_t  = (g_unc["oxy"]   * dp).sum("pres", skipna=True) / dp.sum("pres")
unc_bar_t = (g_unc["uncer"] * dp).sum("pres", skipna=True) / dp.sum("pres")

# time-mean fields
o2_bar  = o2_bar_t.mean("time", skipna=True)
unc_bar = unc_bar_t.mean("time", skipna=True)

eps = np.finfo(float).eps
rel_unc_pct = (unc_bar / (np.abs(o2_bar) + eps)) * 100.0  # percent

# threshold over finite points only
vals = rel_unc_pct.values
thr = np.nanpercentile(vals[np.isfinite(vals)], PCTL_UNC)

high_unc = (rel_unc_pct > thr)
high_unc = high_unc.where(np.isfinite(gtrend))  # only where trend exists

print(f"GOBAI relative-uncertainty threshold (upper 10%): {thr:.2f}%")

# ----------------------------
# 3) Apply Mediterranean mask
# ----------------------------
lon = gtrend["lon"].values
lat = gtrend["lat"].values
lon2d, lat2d = np.meshgrid(lon, lat)

med = mask_mediterranean(lat2d, lon2d)
med_da = xr.DataArray(
    med,
    dims=("lat", "lon"),
    coords={"lat": gtrend["lat"], "lon": gtrend["lon"]}
)

sign_disagree = sign_disagree.where(~med_da)
high_unc      = high_unc.where(~med_da)

# ----------------------------
# 4) Build boolean masks for plotting
# ----------------------------
# Convert xarray -> numpy booleans safely
disagree_np = np.asarray(sign_disagree.fillna(False).values, dtype=bool)
highunc_np  = np.asarray(high_unc.fillna(False).values, dtype=bool)

mask_low  = disagree_np & (~highunc_np)
mask_high = disagree_np & (highunc_np)

# ----------------------------
# 4b) Summary stats (counts + percentages)
# ----------------------------
# "valid" means both trends exist (finite) AND not Mediterranean (because you masked those out)
valid_np = np.isfinite(gtrend.values) & np.isfinite(itrend.values) & (~med)

total_valid = int(valid_np.sum())

# agree/disagree within valid
total_disagree = int(disagree_np.sum())
total_agree    = int(total_valid - total_disagree)

# high-unc among disagree
total_disagree_high = int(mask_high.sum())
total_disagree_low  = int(mask_low.sum())

print("\n===== Pixel statistics (excluding Mediterranean) =====")
print(f"Total valid pixels (both trends finite): {total_valid}")

if total_valid > 0:
    print(f"Agree pixels:    {total_agree} ({100*total_agree/total_valid:.2f}%)")
    print(f"Disagree pixels: {total_disagree} ({100*total_disagree/total_valid:.2f}%)")
else:
    print("No valid pixels found (check masks / NaNs).")

if total_disagree > 0:
    print(f"Disagree pixels w/ HIGH GOBAI unc: {total_disagree_high} "
          f"({100*total_disagree_high/total_disagree:.2f}%)")
    print(f"Disagree pixels w/ low/mod unc:    {total_disagree_low} "
          f"({100*total_disagree_low/total_disagree:.2f}%)")
else:
    print("No disagreement pixels found.")
print("======================================================\n")

# (Optional) keep your original quick print too:
N  = total_disagree
Nh = total_disagree_high
print(f"Total disagreement pixels: {N}")
if N > 0:
    print(f"...of which high-uncertainty: {Nh} ({100*Nh/N:.1f}%)")

# ----------------------------
# 5) Plot
# ----------------------------
fig = plt.figure(figsize=(12, 6), dpi=300)
ax = fig.add_subplot(1, 1, 1, projection=ccrs.Robinson())
ax.set_global()

land = cfeature.NaturalEarthFeature(
    "physical", "land", "50m",
    edgecolor="none", facecolor="lightgrey"
)
ax.add_feature(land)

gl = ax.gridlines(draw_labels=True, alpha=0.3, linestyle="--")
gl.top_labels = gl.bottom_labels = True
gl.left_labels = gl.right_labels = True
gl.xformatter = cticker.LongitudeFormatter()
gl.yformatter = cticker.LatitudeFormatter()
fig.canvas.draw()
for lbl in gl.xlabel_artists + gl.ylabel_artists:
    lbl.set_fontsize(6)

# Disagreement, low/moderate uncertainty (black)
ax.scatter(
    lon2d[mask_low], lat2d[mask_low],
    s=DOTSIZE, color="black",
    transform=ccrs.PlateCarree(),
    label="Sign disagreement (low/moderate GOBAI uncertainty)"
)

# Disagreement, high uncertainty (red)
ax.scatter(
    lon2d[mask_high], lat2d[mask_high],
    s=DOTSIZE, color="red",
    transform=ccrs.PlateCarree(),
    label="Sign disagreement (high GOBAI uncertainty)"
)

ax.set_title(
    "Sign disagreement of thickness-weighted mean O$_2$ trends (GOBAI vs ML4O2)\n"
    "0–1000 m, 2004–2020; colored by GOBAI relative uncertainty (upper 10%)",
    fontsize=11
)

# ax.legend(loc="lower left", fontsize=8, frameon=False)

outpath = "figs/FigS_O2_sign_disagreement_uncertainty.png"
fig.savefig(outpath, bbox_inches="tight", dpi=300)
plt.show()

print("Saved:", outpath)
```
